```python
# fsLR-32k length
n_64k = pet_raw.shape[1]
n_32k = int(n_64k/2)

# Calculate the zscore
hc = df['group'].values=="Healthy"
mu = np.nanmean(pet_raw[hc, :], axis=0)
sd = np.nanstd(pet_raw[hc, :], axis=0, ddof=1)
pet_z = (pet_raw - mu) / sd

# Flip R >> L 18F-mk6240 zscores
mkz_ipsi, mkz_contra = flip_mat(pet_z[:,0:n_32k], pet_z[:,n_32k:n_64k], df['lateralization'].values, flip='R')

# Merge ipsi and contra
tauMK6240z = np.concatenate((mkz_ipsi, mkz_contra), axis=1)

# Flip R >> L 18F-mk6240
mk_ipsi, mk_contra = flip_mat(pet_raw[:,0:n_32k], pet_raw[:,n_32k:n_64k], df['lateralization'].values, flip='R')

# Merge ipsi and contra
tauMK6240 = np.concatenate((mk_ipsi, mk_contra), axis=1)
```

# Sex analysis

```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd

def plot_mk6240_group_summary_by_sex(slm, mk_matched, df, threshold=0.01, png_file=None):
    """
    Compute and plot group-level MK6240 SUVR summary divided by sex
    for ipsilateral and contralateral hemispheres, and run ANOVA tests.

    Added:
        - ANOVA for ipsilateral and contralateral
        - Summary table including mean ± SD, F, df, p-value
    """

    # Copy p-values and define hemispheric split
    pvalues = np.copy(slm.P["pval"]["C"])
    n_64k = mk_matched.shape[1]
    n_32k = n_64k // 2

    mk_ipsi = mk_matched[:, :n_32k]
    mk_contra = mk_matched[:, n_32k:]

    # Binarize significant vertices
    p_bin = (pvalues < threshold).astype(int)
    p_bin_ipsi = p_bin[:n_32k]
    p_bin_contra = p_bin[n_32k:]

    # Compute SUVR features
    df = df.copy()
    df['mk6240.sig.ipsi'] = np.mean(mk_ipsi[:, p_bin_ipsi == 1], axis=1)
    df['mk6240.sig.contra'] = np.mean(mk_contra[:, p_bin_contra == 1], axis=1)
    df['mk6240.mean.ipsi'] = np.mean(mk_ipsi, axis=1)
    df['mk6240.mean.contra'] = np.mean(mk_contra, axis=1)

    # Define combined group-sex label
    df['group_sex'] = df['group'] + "-" + df['sex']

    # Order and palette
    order = ["Healthy-F", "Healthy-M", "Patient-F", "Patient-M"]
    palette = {
        "Healthy-F": "#666666",
        "Healthy-M": "#999999",
        "Patient-F": "#ff7777",
        "Patient-M": "#ff4444"
    }

    # ----------------------
    # ANOVA SUMMARY TABLE
    
    features = ["mk6240.sig.ipsi", "mk6240.sig.contra"]
    names = ["Ipsilateral", "Contralateral"]
    
    summary_rows = []
    posthoc_results = {}
    
    for feat, name in zip(features, names):
    
        # Prepare groups in fixed order
        groups = [df[df['group_sex'] == g][feat].dropna() for g in order]
    
        # -------------------------
        # ANOVA
        # -------------------------
        F, p = f_oneway(*groups)
        df_between = len(groups) - 1
        df_within = df.shape[0] - len(groups)
    
        # group-level summary
        means = df.groupby("group_sex")[feat].mean()
        sds   = df.groupby("group_sex")[feat].std()
    
        # populate summary table row
        row = {
            "Measure": name,
            "Healthy-F mean±sd": f"{means['Healthy-F']:.3f} ± {sds['Healthy-F']:.3f}",
            "Healthy-M mean±sd": f"{means['Healthy-M']:.3f} ± {sds['Healthy-M']:.3f}",
            "Patient-F mean±sd": f"{means['Patient-F']:.3f} ± {sds['Patient-F']:.3f}",
            "Patient-M mean±sd": f"{means['Patient-M']:.3f} ± {sds['Patient-M']:.3f}",
            "F-stat": f"{F:.3f}",
            "df": f"{df_between}, {df_within}",
            "p-value": f"{p:.4g}"
        }
        summary_rows.append(row)
    
        # -------------------------
        # POST-HOC: Tukey HSD
        # -------------------------
        data = df[[feat, "group_sex"]].dropna()
    
        tukey = pairwise_tukeyhsd(
            endog=data[feat],
            groups=data["group_sex"],
            alpha=0.05
        )
    
        # Store the summary object
        posthoc_results[name] = tukey.summary()
    
    # Build summary table
    summary_table = pd.DataFrame(summary_rows)
    
    # -------------------------
    # Convert Tukey results → DataFrame
    # -------------------------
    
    posthoc_frames = []
    
    for name, table in posthoc_results.items():
    
        # Convert statsmodels SimpleTable → pandas DataFrame
        df_post = pd.DataFrame(
            table.data[1:],       # skip header row
            columns=table.data[0]
        )
        
        df_post.insert(0, "Measure", name)  # tag ipsi/contra
        posthoc_frames.append(df_post)
    
    posthoc_df = pd.concat(posthoc_frames, ignore_index=True)

    # ----------------------
    # Plotting
    ymin = df[["mk6240.sig.ipsi", "mk6240.sig.contra"]].min().min()
    ymax = df[["mk6240.sig.ipsi", "mk6240.sig.contra"]].max().max()
    buffer = (ymax - ymin) * 0.05

    plt.figure(figsize=(10, 3))

    for i, (feature, title) in enumerate(zip(features, names), 1):
        plt.subplot(1, 2, i)

        sns.stripplot(
            data=df,
            x='group_sex',
            y=feature,
            order=order,
            hue='group_sex',
            palette=palette,
            alpha=0.6,
            jitter=True,
            dodge=False
        )

        # Overlay mean ± SD per group
        for x_pos, label in enumerate(order):
            g = df[df['group_sex'] == label][feature]
            if len(g) > 0:
                mean_val = g.mean()
                sd_val = g.std()
                plt.plot(x_pos, mean_val, 'o', color='black')
                plt.vlines(x_pos, mean_val - sd_val, mean_val + sd_val,
                           color='black', linewidth=2)

        plt.title(title, fontsize=13)
        plt.xlabel('')
        plt.ylabel('SUVR')
        plt.ylim(ymin - buffer, ymax + buffer)
        plt.xticks(rotation=25)
        sns.despine()

    plt.tight_layout()
    
    if png_file is not None:
        plt.savefig(png_file, format='svg', dpi=300, bbox_inches='tight', transparent=False)
        
    plt.show()

    display(summary_table)
    display(posthoc_df)
    
```

## Interaction: difference in group effect (Female > Male)

```python
# Set the terms of the model
term_grp = FixedEffect(df['group'])
term_age = FixedEffect(df['age'])
term_sex = FixedEffect(df['sex'])
term_subject = MixedEffect(df['participant_id'])

# boolean masks
is_healthy = (df.group == 'Healthy')
is_patient = (df.group == 'Patient')
is_male = (df.sex == 'M')
is_female = (df.sex == 'F')

# Model is is mixed Subject is set as random Mixed effect variable
print("Y ~ group + sex + group:sex + age + 1/subject")
model = term_grp + term_sex + term_grp*term_sex + term_age  + term_subject

# Group effect IN MALES  -> (Healthy_male) - (Patient_male)
contrast_group_male = (is_patient & is_male).astype(int) - (is_healthy & is_male).astype(int)


# Group effect IN FEMALES -> (Healthy_female) - (Patient_female)
contrast_group_female = (is_patient & is_female).astype(int) - (is_healthy & is_female).astype(int)

# Interaction: difference in group effect (Female minus Male)
contrast_interaction = contrast_group_female - contrast_group_male

# Model is is mixed Subject is set as random Mixed effect variable
print("Y ~ group + sex + group:sex + age + 1/subject")
model = term_grp + term_sex + term_grp*term_sex + term_age  + term_subject

# fitting the model
slm_mixed = SLM(
        model,
        contrast_interaction,
        mask=fslr32k_mask,
        surf=fslr32k,
        correction="rft",
        two_tailed=True,
        cluster_threshold=0.025
    )
slm_mixed.fit(tauMK6240z)

# Plot t-values
plot_ctx_slm(fslr32k_lh, fslr32k_rh, slm_mixed, color_range=(-3,3), Str='Interaction F>M',  Col="RdBu_r", mask=fslr32k_mask,
             scale=fig_scale, Save=save_fig, png_file=f"{osf_path}/Sup_sex-F>M_tmap.png")
```

```python
# p>0.025 Plot P-values
plot_ctx_pval(fslr32k_lh, fslr32k_rh, slm_mixed, Str='p-values', Col="inferno", Thr=0.025, 
             scale=fig_scale, Save=save_fig, png_file=f"{osf_path}/Sup_sex-F>M_pvals.png")
```

```python
def sort_and_mask(matrix, labels, mask):
    """
    matrix : 2D array (subjects × vertices)
    labels : 1D array-like, same #rows as matrix
    mask   : boolean vector (length = vertices), True = masked vertex
    """

    import numpy as np

    # Compute per-row means excluding masked vertices
    row_mean = np.nanmean(matrix[:, mask], axis=1)

    # Sort by mean (descending)
    order = np.argsort(row_mean)[::-1]

    # Apply sorting
    matrix_sorted = matrix[order, :].copy()
    labels_sorted = np.array(labels)[order]

    # Mask → nan
    matrix_sorted[:, ~mask] = np.nan

    # Plot the array
    N = matrix_sorted.shape[0]
    img = plot_hemispheres(fslr32k_lh, fslr32k_rh, array_name=matrix_sorted, size=(600, 88*N), cmap='cmo.amp',
                     embed_nb=True, label_text={'left':labels_sorted}, color_bar='right', color_range=(0.2,2.2),
                     zoom=1.5, nan_color=[0,0,0,1], scale=2)
    return(img)
```
